# Display Cluster

This tool visualizes a cluster of pieces. It can be used to display both the ground truth assembly and the assembly generated by our method.

```python
%load_ext autoreload
%autoreload 2
```

```python
from piece_assemble.cluster import Cluster, DummyClusterScorer
from piece_assemble.piece import Piece
from piece_assemble.load import load_images
from piece_assemble.image import np_to_pil
import json
```

```python
def load_cluster(cluster_dict_path: str, img_dir_path: str, scale: int = 1) -> Cluster:
    with open(cluster_dict_path) as f:
        cluster_dict = json.load(f)
    
    img_ids, imgs, masks = load_images(img_dir_path, scale)
    pieces = [
        Piece.from_image(
            img_name, img, mask
        )
        for img_name, img, mask in zip(img_ids, imgs, masks)
    ]
    pieces_dict = {piece.name: piece for piece in pieces}

    return Cluster.from_dict(cluster_dict, pieces_dict, DummyClusterScorer(), **Cluster.get_default_config()), pieces_dict
    
```

## Inputs

- `cluster_dict_path`: Path to a JSON file containing the cluster description. This can be either the ground truth file (pieces.json) generated by our synthetic puzzle generator, or the output of our assembly algorithm (produced by running run_assembly.py).

- `img_dir_path`: Directory containing all piece images.

- `scale`: Scaling factor for the piece images. It should match the scale used during prediction. When displaying the ground truth solution, set this value to 1.

```python
cluster_dict_path = ""
img_dir_path = ""
scale = 1

cluster, pieces_dict = load_cluster(cluster_dict_path, img_dir_path, scale)
```

```python
np_to_pil(cluster.draw(draw_contours=False, draw_borders=False, thickness=2))
```
